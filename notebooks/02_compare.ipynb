Puts the metrics of analysed experiments side by side, shows the parameter differences,
lists per-sequence F differences and exports everything to Excel.

The primary metrics are **mAP, AP50 and AP75**; per-sequence differences are listed for each of them.
F at the F-max threshold (`F_opt`, every experiment at its own best threshold as in the VOT-LT protocol) and
P / R / F1 at the fixed threshold (`eval_score_thr`) are also in the tables.
